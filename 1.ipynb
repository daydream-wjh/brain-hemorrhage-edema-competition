```python
import pandas as pd

# 读取数据集
df = pd.read_excel('表1-患者列表及临床信息.xlsx')

# 删除ID列和首次检查流水号列
df.drop(['ID', '首次检查流水号'], axis=1, inplace=True)

# 删除HM_volume和ED_volume中的异常值
df = df[(df['HM_volume'] > 0) & (df['ED_volume'] > 0)]

# 将HM_ACA_R_Ratio、HM_MCA_R_Ratio、HM_PCA_R_Ratio、HM_Pons_Medulla_R_Ratio、HM_Cerebellum_R_Ratio、HM_ACA_L_Ratio、HM_MCA_L_Ratio、HM_PCA_L_Ratio、HM_Pons_Medulla_L_Ratio和HM_Cerebellum_L_Ratio中的异常值替换为0
df.loc[(df['HM_ACA_R_Ratio'] < 0) | (df['HM_ACA_R_Ratio'] > 1), 'HM_ACA_R_Ratio'] = 0
df.loc[(df['HM_MCA_R_Ratio'] < 0) | (df['HM_MCA_R_Ratio'] > 1), 'HM_MCA_R_Ratio'] = 0
df.loc[(df['HM_PCA_R_Ratio'] < 0) | (df['HM_PCA_R_Ratio'] > 1), 'HM_PCA_R_Ratio']=0
```
